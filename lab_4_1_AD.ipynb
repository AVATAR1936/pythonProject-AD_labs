```python
import pandas as pd
df = pd.read_csv("household_power_consumption.txt", sep=';', na_values='?', low_memory=False)
df.head()
```

```python
import numpy as np

types = [
    ("Date", "U10"), 
    ("Time", "U8"), 
    ("Global_active_power", "float64"),
    ("Global_reactive_power", "float64"), 
    ("Voltage", "float64"),
    ("Global_intensity", "float64"), 
    ("Sub_metering_1", "float64"),
    ("Sub_metering_2", "float64"), 
    ("Sub_metering_3", "float64")
]

data = np.genfromtxt(
    "household_power_consumption.txt",
    delimiter=';',
    dtype=types,
    names=True,
    missing_values=["?", "nan"],
    filling_values=np.nan,
    encoding="utf-8"
)

print(data[:5])
```

```python
df_after_18 = df[df['Time'] > '18:00:00']
print(df_after_18.head())

df_after_18 = df_after_18[df_after_18['Global_active_power'] > 6]
print(df_after_18.head())

df_main_group_2 = df_after_18[
    (df_after_18['Sub_metering_2'] > df_after_18['Sub_metering_1']) & (df_after_18['Sub_metering_2'] > df_after_18['Sub_metering_3'])
]
print(df_main_group_2.head())

# Ділення на половини
half = len(df_main_group_2) // 2
first_half = df_main_group_2.iloc[:half]
second_half = df_main_group_2.iloc[half:]

# Кожен третій із першої половини
selected_first = first_half.iloc[::3]

# Кожен четвертий із другої половини
selected_second = second_half.iloc[::4]

final_result = pd.concat([selected_first, selected_second])

print(final_result.head())
print(f"\nУсього відібрано записів: {len(final_result)}")
```

```python
after_18_mask = data['Time'] > '18:00:00'

power_mask = data['Global_active_power'] > 6

group_mask = (data['Sub_metering_2'] > data['Sub_metering_1']) & (data['Sub_metering_2'] > data['Sub_metering_3'])

# Об’єднання всіх умов
final_mask = after_18_mask & power_mask & group_mask
filtered = data[final_mask]

half = len(filtered) // 2
first_half = filtered[:half]
second_half = filtered[half:]

every_third = first_half[::3]

every_fourth = second_half[::4]

final_result = np.concatenate((every_third, every_fourth))

print("Перші 5 записів:")
print(final_result[:5])
print(f"\nЗагальна кількість відібраних записів: {len(final_result)}")
```
